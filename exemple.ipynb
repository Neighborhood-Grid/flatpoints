```python
import numpy as np
import src.flatpoints as fp
```

# Load example 
3D point cloud on a surface


```python
points = np.load("data/cow_points.npy")   
print(points.shape)
```

```python
single_face = False #the cow is topologicaly closed, so we want to build left and right surfaces (not a single surface to avoid overlaps or tears)
res = fp.flatten(points, single_face = single_face)

np.set_printoptions(threshold=0, edgeitems=0)
print(res)
```

# Isomaps (x, y, z) -> (u, v). 
Left and right side of the surface

```python
_ = fp.plot(res, points)
```

# Pixel layout [i, j] -> point 
RGB = (x, y, z) coordinates

```python
_ = fp.imshow(res, points)
```
